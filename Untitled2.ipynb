```python
from pathlib import Path

import torch
import torch.nn as nn
import torch.nn.functional as F

from torchvision import transforms
from torchvision.models import efficientnet_b4

from PIL import Image
import matplotlib.pyplot as plt

print("PyTorch version:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Using device:", device)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

```python
MODEL_PATH = Path("model/best_efficientnet_b4.pth")

checkpoint = torch.load(
    MODEL_PATH,
    map_location=device
)

class_names = checkpoint["class_names"]

NUM_CLASSES = len(class_names)

print("Number of classes:", NUM_CLASSES)

print("\nClasses:")
for i, class_name in enumerate(class_names):
    print(i, "→", class_name)
```

```python
model = efficientnet_b4(weights=None)

model.classifier[1] = nn.Linear(
    1792,
    NUM_CLASSES
)

model = model.to(device)

print("EfficientNet-B4 created.")
```

```python
model.load_state_dict(
    checkpoint["model_state_dict"]
)

model.eval()

print("✓ Trained model loaded successfully")
print("Checkpoint epoch:", checkpoint["epoch"])
print(
    "PlantVillage validation Macro-F1:",
    checkpoint["macro_f1"]
)
```

```python
test_transform = transforms.Compose([
    transforms.Resize(420),
    transforms.CenterCrop(380),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])
```

```python
TEST_DIR = Path("data/test_images")

image_extensions = {
    ".jpg", ".jpeg", ".png",
    ".JPG", ".JPEG", ".PNG"
}

test_images = sorted([
    p for p in TEST_DIR.iterdir()
    if p.is_file() and p.suffix in image_extensions
])

print("Images found:", len(test_images))

for image_path in test_images:
    print(" -", image_path.name)
```

```python
results = []

for image_path in test_images:

    # Load image
    image = Image.open(image_path).convert("RGB")

    # Apply the exact validation preprocessing
    input_tensor = test_transform(image).unsqueeze(0).to(device)

    # Prediction
    with torch.no_grad():
        with torch.autocast(
            device_type=device.type,
            dtype=torch.float16,
            enabled=(device.type == "cuda")
        ):
            outputs = model(input_tensor)

        probabilities = F.softmax(outputs, dim=1)

    # Top 5 predictions
    top5_prob, top5_indices = torch.topk(
        probabilities,
        k=5,
        dim=1
    )

    top5_prob = top5_prob[0].cpu().numpy()
    top5_indices = top5_indices[0].cpu().numpy()

    results.append({
        "image": image_path.name,
        "prediction": class_names[top5_indices[0]],
        "confidence": float(top5_prob[0]),
        "top5": [
            (class_names[idx], float(prob))
            for idx, prob in zip(top5_indices, top5_prob)
        ]
    })

print(f"✓ Predictions completed for {len(results)} images.")
```

```python
for result in results:

    print("=" * 80)
    print("IMAGE:", result["image"])
    print("=" * 80)

    print(
        f"Prediction : {result['prediction']}"
    )

    print(
        f"Confidence : {result['confidence'] * 100:.2f}%"
    )

    print("\nTop 5 predictions:")

    for rank, (class_name, probability) in enumerate(
        result["top5"],
        start=1
    ):
        print(
            f"{rank}. {class_name:<55} "
            f"{probability * 100:.2f}%"
        )

    print()
```

```python
for i, result in enumerate(results, start=1):

    print(f"{i:02d}. {result['image']}")
    print(f"    Prediction : {result['prediction']}")
    print(f"    Confidence : {result['confidence'] * 100:.2f}%")
    print("-" * 80)
```

```python
def pretty_class_name(class_name):
    crop, disease = class_name.split("___", 1)
    disease = disease.replace("_", " ")
    return f"{crop} — {disease}"


for i, result in enumerate(results, start=1):

    print(f"{i:02d}. {result['image']}")
    print(f"    🌱 Prediction : {pretty_class_name(result['prediction'])}")
    print(f"    📊 Confidence : {result['confidence'] * 100:.2f}%")
    print("-" * 80)
```

```python
# Test images from test_2 folder

test_2_dir = Path("data/test_2")
test_2_images = sorted(
    [p for p in test_2_dir.iterdir()
     if p.suffix.lower() in [".jpg", ".jpeg", ".png"]]
)

print("Images found:", len(test_2_images))
```

```python
# Test test_2 images - prediction and confidence only

for image_path in test_2_images:
    image = Image.open(image_path).convert("RGB")
    
    input_tensor = test_transform(image).unsqueeze(0).to(device)
    
    with torch.no_grad():
        with torch.autocast(
            device_type=device.type,
            dtype=torch.float16,
            enabled=(device.type == "cuda")
        ):
            outputs = model(input_tensor)
        
        probabilities = F.softmax(outputs, dim=1)
        top_prob, top_index = torch.max(probabilities, dim=1)
    
    prediction = class_names[top_index.item()]
    confidence = top_prob.item() * 100
    
    print(f"{image_path.name}")
    print(f"Prediction: {prediction}")
    print(f"Confidence: {confidence:.2f}%")
    print("-" * 60)
```
