```python
from collections import Counter

image_extensions = {".jpg", ".jpeg", ".png", ".bmp", ".JPG", ".JPEG", ".PNG", ".BMP"}

class_counts = {}

for class_folder in COLOR_PATH.iterdir():
    if class_folder.is_dir():
        count = sum(
            1 for file in class_folder.iterdir()
            if file.is_file() and file.suffix in image_extensions
        )
        class_counts[class_folder.name] = count

for class_name, count in sorted(class_counts.items()):
    print(f"{class_name:65} {count}")

print("\nTotal classes:", len(class_counts))
print("Total images:", sum(class_counts.values()))
```

```python
import matplotlib.pyplot as plt

# Sort classes by number of images
sorted_classes = sorted(class_counts.items(), key=lambda x: x[1])

classes = [x[0] for x in sorted_classes]
counts = [x[1] for x in sorted_classes]

plt.figure(figsize=(12, 10))
plt.barh(classes, counts)

plt.xlabel("Number of Images")
plt.ylabel("Class")
plt.title("PlantVillage Class Distribution")

plt.tight_layout()
plt.show()
```

```python
from collections import Counter
from PIL import Image

dimension_counts = Counter()
corrupted_images = []

for class_folder in COLOR_PATH.iterdir():

    if not class_folder.is_dir():
        continue

    for file in class_folder.iterdir():

        if file.is_file() and file.suffix.lower() in image_extensions:

            try:
                with Image.open(file) as img:
                    dimension_counts[img.size] += 1

            except Exception:
                corrupted_images.append(str(file))

print("Most common image dimensions:\n")

for dimension, count in dimension_counts.most_common(15):
    print(f"{dimension}: {count}")

print("\nTotal unique dimensions:", len(dimension_counts))
print("Corrupted/unreadable images:", len(corrupted_images))
```

```python
 import hashlib
from collections import defaultdict

hash_to_files = defaultdict(list)

for class_folder in COLOR_PATH.iterdir():
    if not class_folder.is_dir():
        continue

    for file in class_folder.iterdir():
        if file.is_file() and file.suffix.lower() in image_extensions:
            with open(file, "rb") as f:
                file_hash = hashlib.md5(f.read()).hexdigest()

            hash_to_files[file_hash].append(str(file))

duplicates = {
    h: files
    for h, files in hash_to_files.items()
    if len(files) > 1
}

print("Total images checked:", len(hash_to_files))
print("Duplicate groups:", len(duplicates))

if duplicates:
    print("\nExample duplicate groups:")
    for files in list(duplicates.values())[:5]:
        print(files)
```

```python
from pathlib import Path
import pandas as pd
import hashlib
from sklearn.model_selection import train_test_split

# Dataset location
DATASET_PATH = Path("data/archive/plantvillage dataset/color")

# Find all images
image_extensions = {".jpg", ".jpeg", ".png", ".JPG", ".JPEG", ".PNG"}

records = []

for class_folder in sorted(DATASET_PATH.iterdir()):
    if not class_folder.is_dir():
        continue

    class_name = class_folder.name

    for image_path in class_folder.iterdir():
        if image_path.suffix in image_extensions:
            records.append({
                "image_path": str(image_path),
                "class_name": class_name
            })

# Create DataFrame
df = pd.DataFrame(records)

# Create numeric labels
class_names = sorted(df["class_name"].unique())
class_to_idx = {name: idx for idx, name in enumerate(class_names)}

df["label"] = df["class_name"].map(class_to_idx)

print("Total images:", len(df))
print("Total classes:", len(class_names))

df.head()
```

```python
print("Classes:")
for idx, name in enumerate(class_names):
    print(f"{idx:2d} -> {name}")

print("\nImages per class:")
print(df["class_name"].value_counts().sort_index())
```

```python
def get_file_hash(path):
    hasher = hashlib.md5()

    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(8192), b""):
            hasher.update(chunk)

    return hasher.hexdigest()


print("Calculating image hashes...")

df["image_hash"] = df["image_path"].apply(get_file_hash)

print("Done.")
print("Unique images:", df["image_hash"].nunique())
print("Duplicate files:", len(df) - df["image_hash"].nunique())
```

```python
# One representative row for each unique image
unique_df = df.drop_duplicates(subset="image_hash").copy()

train_groups, val_groups = train_test_split(
    unique_df,
    test_size=0.20,
    random_state=42,
    stratify=unique_df["label"]
)

# Get the hashes assigned to each split
train_hashes = set(train_groups["image_hash"])
val_hashes = set(val_groups["image_hash"])

# Assign every original file according to its image hash
df["split"] = df["image_hash"].apply(
    lambda h: "train" if h in train_hashes else "val"
)

print(df["split"].value_counts())
```

```python
train_hashes = set(df.loc[df["split"] == "train", "image_hash"])
val_hashes = set(df.loc[df["split"] == "val", "image_hash"])

overlap = train_hashes.intersection(val_hashes)

print("Duplicate hashes shared between train and validation:", len(overlap))
```

```python
split_distribution = pd.crosstab(
    df["class_name"],
    df["split"]
)

split_distribution["total"] = (
    split_distribution["train"] +
    split_distribution["val"]
)

split_distribution
```

```python
print("Classes missing from training:")
print(
    set(class_names) -
    set(df.loc[df["split"] == "train", "class_name"])
)

print("\nClasses missing from validation:")
print(
    set(class_names) -
    set(df.loc[df["split"] == "val", "class_name"])
)
```

```python
# Save class mapping
class_mapping = pd.DataFrame({
    "label": range(len(class_names)),
    "class_name": class_names
})

# Save CSV files
df[df["split"] == "train"][
    ["image_path", "class_name", "label"]
].to_csv("data/train.csv", index=False)

df[df["split"] == "val"][
    ["image_path", "class_name", "label"]
].to_csv("data/val.csv", index=False)

class_mapping.to_csv("data/class_mapping.csv", index=False)

print("Saved:")
print("  data/train.csv")
print("  data/val.csv")
print("  data/class_mapping.csv")
```

```python
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

from torchvision import transforms
from PIL import Image

import pandas as pd
import numpy as np
from pathlib import Path
```

```python
class PlantDiseaseDataset(Dataset):

    def __init__(self, dataframe, transform=None):
        self.dataframe = dataframe.reset_index(drop=True)
        self.transform = transform

    def __len__(self):
        return len(self.dataframe)

    def __getitem__(self, index):

        row = self.dataframe.iloc[index]

        image_path = row["image_path"]
        label = int(row["label"])

        # Load image
        image = Image.open(image_path).convert("RGB")

        # Apply transformations
        if self.transform:
            image = self.transform(image)

        return image, label
```

```python
train_transform = transforms.Compose([
    transforms.RandomResizedCrop(
        size=380,
        scale=(0.8, 1.0)
    ),
    
    transforms.RandomHorizontalFlip(
        p=0.5
    ),
    
    transforms.RandomRotation(
        degrees=15
    ),
    
    transforms.ColorJitter(
        brightness=0.2,
        contrast=0.2,
        saturation=0.2,
        hue=0.05
    ),
    
    transforms.ToTensor(),
    
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


val_transform = transforms.Compose([
    transforms.Resize(420),
    
    transforms.CenterCrop(380),
    
    transforms.ToTensor(),
    
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])
```

```python
train_dataset = PlantDiseaseDataset(
    train_df,
    transform=train_transform
)

val_dataset = PlantDiseaseDataset(
    val_df,
    transform=val_transform
)

print("Training samples:", len(train_dataset))
print("Validation samples:", len(val_dataset))
```

```python
image, label = train_dataset[0]

print("Image shape:", image.shape)
print("Label:", label)
print("Class:", class_names[label])
print("Data type:", image.dtype)
```

```python
BATCH_SIZE = 8

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

print("Train batches:", len(train_loader))
print("Validation batches:", len(val_loader))
```

```python
images, labels = next(iter(train_loader))

print("Batch image shape:", images.shape)
print("Batch labels shape:", labels.shape)

print("Moving batch to GPU...")

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

images = images.to(device)
labels = labels.to(device)

print("Images device:", images.device)
print("Labels device:", labels.device)
```
