```python
from pathlib import Path

import torch
import torch.nn as nn
from torchvision import transforms
from torchvision.models import efficientnet_b4

from PIL import Image
import pandas as pd
import numpy as np

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix
)
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Device:", device)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

```python
checkpoint_path = Path("model/best_efficientnet_b4.pth")

checkpoint = torch.load(
    checkpoint_path,
    map_location=device,
    weights_only=False
)

class_names = checkpoint["class_names"]
NUM_CLASSES = len(class_names)

print("Best checkpoint epoch:", checkpoint["epoch"])
print("Best PlantVillage Macro-F1:", checkpoint["macro_f1"])
print("Number of classes:", NUM_CLASSES)
```

```python
model = efficientnet_b4(weights=None)

model.classifier[1] = nn.Linear(
    model.classifier[1].in_features,
    NUM_CLASSES
)

model.load_state_dict(checkpoint["model_state_dict"])

model = model.to(device)
model.eval()

print("Epoch-5 model loaded successfully.")
```

```python
test_transform = transforms.Compose([
    transforms.Resize(420),
    transforms.CenterCrop(380),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])
```

```python
plantdoc_df = pd.read_csv("data/plantdoc_test.csv")

print("PlantDoc test images:", len(plantdoc_df))
print(plantdoc_df.head())
```

```python
predictions = []
actual_labels = []

for _, row in plantdoc_df.iterrows():

    image_path = Path(row["image_path"])

    image = Image.open(image_path).convert("RGB")

    input_tensor = test_transform(image).unsqueeze(0).to(device)

    with torch.no_grad():

        with torch.autocast(
            device_type=device.type,
            dtype=torch.float16,
            enabled=(device.type == "cuda")
        ):
            outputs = model(input_tensor)

        probabilities = torch.softmax(outputs, dim=1)

        confidence, predicted_index = torch.max(
            probabilities,
            dim=1
        )

    predictions.append(predicted_index.item())
    actual_labels.append(row["label"])

print("Predictions completed:", len(predictions))
```

```python
accuracy = accuracy_score(
    actual_labels,
    predictions
)

macro_f1 = f1_score(
    actual_labels,
    predictions,
    average="macro"
)

print(f"PlantDoc Accuracy: {accuracy:.4f} ({accuracy * 100:.2f}%)")
print(f"PlantDoc Macro-F1: {macro_f1:.4f} ({macro_f1 * 100:.2f}%)")
```

```python
print(
    classification_report(
        actual_labels,
        predictions,
        labels=range(NUM_CLASSES),
        target_names=class_names,
        zero_division=0
    )
)
```

```python
test_1_dir = Path("data/test_images")
test_2_dir = Path("data/test_2")

test_1_images = sorted([
    p for p in test_1_dir.iterdir()
    if p.suffix.lower() in [".jpg", ".jpeg", ".png"]
])

test_2_images = sorted([
    p for p in test_2_dir.iterdir()
    if p.suffix.lower() in [".jpg", ".jpeg", ".png"]
])

print("Test 1 images:", len(test_1_images))
print("Test 2 images:", len(test_2_images))
```

```python
def predict_image(image_path):
    image = Image.open(image_path).convert("RGB")
    
    input_tensor = test_transform(image).unsqueeze(0).to(device)
    
    with torch.no_grad():
        with torch.autocast(
            device_type=device.type,
            dtype=torch.float16,
            enabled=(device.type == "cuda")
        ):
            outputs = model(input_tensor)
        
        probabilities = torch.softmax(outputs, dim=1)
        confidence, predicted_index = torch.max(
            probabilities, dim=1
        )
    
    prediction = class_names[predicted_index.item()]
    confidence = confidence.item() * 100
    
    return prediction, confidence
```

```python
print("=" * 70)
print("TEST 1 — data/test_images")
print("=" * 70)

for i, image_path in enumerate(test_1_images, start=1):
    prediction, confidence = predict_image(image_path)
    
    print(f"\n{i}. Image: {image_path.name}")
    print(f"   Prediction: {prediction}")
    print(f"   Confidence: {confidence:.2f}%")
```

```python
print("=" * 70)
print("TEST 2 — data/test_2")
print("=" * 70)

for i, image_path in enumerate(test_2_images, start=1):
    prediction, confidence = predict_image(image_path)
    
    print(f"\n{i}. Image: {image_path.name}")
    print(f"   Prediction: {prediction}")
    print(f"   Confidence: {confidence:.2f}%")
```
