```python
from pathlib import Path

import torch
import torch.nn as nn
from torchvision import transforms
from torchvision.models import efficientnet_b4

from PIL import Image

import pandas as pd
import numpy as np

from sklearn.metrics import accuracy_score, f1_score, classification_report, confusion_matrix
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Device:", device)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

```python
model_path = Path("model/efficientnet_b4_epoch1.pth")

checkpoint = torch.load(
    model_path,
    map_location=device,
    weights_only=False
)

class_names = checkpoint["class_names"]
NUM_CLASSES = len(class_names)

print("Model:", model_path)
print("Number of classes:", NUM_CLASSES)
print("Saved epoch:", checkpoint.get("epoch", "Not available"))
```

```python
model = efficientnet_b4(weights=None)

model.classifier[1] = nn.Linear(
    model.classifier[1].in_features,
    NUM_CLASSES
)

model.load_state_dict(checkpoint["model_state_dict"])

model = model.to(device)
model.eval()

print("✅ Retrained Epoch-1 model loaded successfully!")
```

```python
test_transform = transforms.Compose([
    transforms.Resize(420),
    transforms.CenterCrop(380),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])
```

```python
plantdoc_df = pd.read_csv("data/plantdoc_test.csv")

print("PlantDoc test images:", len(plantdoc_df))
print()
print(plantdoc_df.head())
```

```python
print("Missing values:")
print(plantdoc_df.isnull().sum())

print("\nNumber of classes in PlantDoc:")
print(plantdoc_df["class_name"].nunique())
```

```python
all_predictions = []
all_true_labels = []

for _, row in plantdoc_df.iterrows():

    image_path = Path(row["image_path"])

    image = Image.open(image_path).convert("RGB")

    input_tensor = test_transform(image).unsqueeze(0).to(device)

    with torch.inference_mode():
        outputs = model(input_tensor)
        prediction = torch.argmax(outputs, dim=1).item()

    all_predictions.append(prediction)
    all_true_labels.append(int(row["label"]))

    del input_tensor, outputs
```

```python
accuracy = accuracy_score(
    all_true_labels,
    all_predictions
)

macro_f1 = f1_score(
    all_true_labels,
    all_predictions,
    average="macro",
    zero_division=0
)

print("PlantDoc Test Results")
print("---------------------")
print(f"Accuracy : {accuracy * 100:.2f}%")
print(f"Macro-F1 : {macro_f1 * 100:.2f}%")
print(f"Correct  : {sum(np.array(all_true_labels) == np.array(all_predictions))}")
print(f"Wrong    : {sum(np.array(all_true_labels) != np.array(all_predictions))}")
```

```python
report = classification_report(
    all_true_labels,
    all_predictions,
    labels=sorted(set(all_true_labels)),
    target_names=[
        class_names[i]
        for i in sorted(set(all_true_labels))
    ],
    zero_division=0
)

print(report)
```
